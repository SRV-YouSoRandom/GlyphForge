```python
from pathlib import Path

project_root = Path.cwd().parent

print(project_root)
```

```python
directories = [
    project_root / "data" / "raw",
    project_root / "data" / "processed",
    project_root / "data" / "benchmark",
    project_root / "results" / "images",
    project_root / "results" / "ocr",
    project_root / "results" / "benchmarks",
    project_root / "src" / "preprocessing",
    project_root / "src" / "compression",
    project_root / "src" / "ocr",
    project_root / "src" / "benchmarking",
]

for directory in directories:
    print(directory, directory.exists())
```

```python
benchmark_root = project_root / "data" / "benchmark"

files = list(benchmark_root.rglob("*"))

image_files = [
    file for file in files
    if file.is_file() and file.suffix.lower() in {".png", ".jpg", ".jpeg", ".webp"}
]

print("Images:", len(image_files))

for image_file in image_files:
    print(image_file.relative_to(project_root))
```

```python
categories = [
    "Real5-OmniDocBench-Scanning",
    "Real5-OmniDocBench-Warping",
    "Real5-OmniDocBench-Screen-Photography",
    "Real5-OmniDocBench-Illumination",
    "Real5-OmniDocBench-Skew",
]

for category in categories:
    category_path = benchmark_root / category
    count = len([
        file for file in category_path.iterdir()
        if file.is_file()
    ])
    print(category, count)
```

```python
clean_image_path = project_root / "data" / "raw" / "clean" / "paddleocr_demo.png"

print(clean_image_path)
print(clean_image_path.exists())
print(clean_image_path.stat().st_size if clean_image_path.exists() else "File not found")
```

```python
image = cv2.imread(str(clean_image_path))

print(image is not None)
print(image.shape)
```

```python
import matplotlib.pyplot as plt

image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(12, 8))
plt.imshow(image_rgb)
plt.axis("off")
plt.show()
```
